# Recall: Classes
In languages like C, you can create data structures but that's mostly it. With python (object oriented lenguage), you can create data structures with their own particula method. The init method is called whenever an object of that class is created; self is used by python (behind the scenes) to point to the instance of the object we are interested in. Here a small example:

```python
class car:
    def __init__(self, model, speed):
        self.model = model
        self.speed = speed

    def speedometer(self):          #if I pass self, I pass all the instance specific variables
        print(f"I am a {self.model} and I am going at {self.speed} km/h")

# Create an object
x = car("porsche", "100")

# Use its method
x.speedometer()
```

# Binary Tree

A Binary Tree Data Structure is a hierarchical data structure in which each node has at most two children, referred to as the left child and the right child. It is commonly used in computer science for efficient storage and retrieval of data, with various operations such as insertion, deletion, and traversal. Some tecn:

Node structure: Each node holds a value and links to up to two children.



*   Root: The top node. Everything descends from it
*   Leaf: A node with no children
*   Height: Longest path from the root to a leaf. This matters because it directly affects performance.


The binary tree is not native in python, we need to define it using classes.









```python
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None


class BinaryTree:
    def __init__(self):
        self.root = None

    def insert(self, value):
        if not self.root:
            self.root = Node(value)
            return

        current = self.root
        while True:
            if value < current.value:
                if current.left is None:
                    current.left = Node(value)
                    return
                current = current.left
            else:
                if current.right is None:
                    current.right = Node(value)
                    return
                current = current.right

```
