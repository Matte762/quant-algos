# Sorting

```python
#Bubble Sort
  #idea: keep swapping so you have local (2 index) minimum
def bubble_sort(arr:list[int]) -> list[int]:
  n = len(arr)
  for i in range(n):
    for j in range(0, n-i-1):
      if arr[j] > arr[j+1]:
        arr[j], arr[j+1] = arr[j+1], arr[j]

  return arr


#Selection Sort
def selection_sort(arr:list[int]) -> list[int]:
  n = len(arr)
  for i in range(n):
    min_index = i
    for j in range(i+1, n):
      if arr[j] < arr[min_index]:
        min_index = j
    arr[i], arr[min_index] = arr[min_index], arr[i]
  return arr

#Slowest Sort (I gave at shot at some recursive kind of sorting)
def slow_sort(arr, i=0):
  n = len(arr)
  if i == n-1:
    return arr

  min = arr[i]
  for s in range(i,n):
    if min > arr[s]:
      min = arr[s]

  if min == arr[i]:
    return slow_sort(arr, i+1)
  else:
    last = arr.pop()
    arr.insert(i, last)
    return slow_sort(arr, i)




import random as rand
#Lucky Sort
def lucky_sort(arr:list[int]) -> list[int]:
  n=len(arr)
  i=0
  j=False
  while j==False:
    i=0
    c=0
    while i<n-1:
      if arr[i]>arr[i+1]:
        c+=1
      i+=1
    if c==0:
      j=True
    else: 
      x=rand.randint(0,n-1)
      y=rand.randint(0,n-1)
      arr[x], arr[y] = arr[y], arr[x]
  return arr

    
  


#Testing - Just for fun

import time
import random
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(267)

algorithms = {
    "bubble": bubble_sort,
    "selection": selection_sort,
    "slow": slow_sort,
    "lucky": lucky_sort,
}

# sizes to test
ns = [10, 11, 12, 13, 20, 30, 100, 200, 300, 1000]

# store times: {algo_name: [t_for_n1, t_for_n2, ...]}
times = {name: [] for name in algorithms.keys()}

for n in ns:
    # base random data as a Python list
    base = np.random.randint(0, n // 2, size=n).tolist()

    print(f"\n=== n = {n} ===")
    for name, func in algorithms.items():
        # Optional: avoid lucky_sort at insane sizes
        if name == "lucky" and n > 13:
            print(f"{name:10s}: skipped (too slow / unbounded)")
            times[name].append(float('nan'))
            continue
      
        arr = base.copy()  

        start = time.perf_counter()
        _ = func(arr)      
        end = time.perf_counter()

        elapsed = round(end - start, 6)
        times[name].append(elapsed)

        print(f"{name:10s}: {elapsed:.6f} s")

# ---- Plotting ----
for name, ts in times.items():
    plt.plot(ns, ts, marker='o', label=name)

plt.xlabel("n (array length)")
plt.ylabel("time (seconds)")
plt.title("Sorting algorithm performance")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

# Determinant of a Matrix nxn

```python
def check_square(mat:list[list[int]]) -> bool:
  rows = len(mat)
  if rows == 0:
    return False
  for i in range(rows):
    if len(mat[i])!=rows:
      return False
  return True

def det(mat:list[list[int]]) -> int:
  rows = len(mat)

  if rows==1:
    return mat[0][0]

  if rows==2:
    return mat[0][0]*mat[1][1]-mat[0][1]*mat[1][0]

  determinant = 0
  for i in range(rows):
    new_mat = [row[:i] + row[i+1:] for row in mat[1:]]
    determinant += ((-1)**i) *mat[0][i]*det(new_mat)

  return determinant




mat = [[1,0,0],[0,1,0],[0,0,1]]

import random as rand
mat = [[rand.randint(0,15) for x in range(11)] for y in range(11)]

if check_square(mat):
  print(mat)
  print("The determinant is " + str(det(mat)))
else:
  print("Not a square matrix")

```

# General operations on Arrays



```python
def remove_dup(nums):
  n=len(nums)
  i=0
  while i<n-1:
    if nums[i]==nums[i+1]:
      nums.pop(i)
      n=n-1
    else:
      i=i+1
  return nums

print(remove_dup([0,0,0,0,0,0,2,0,1,1,1,2,2,3,3,4]))



import math
def divisors(n):
  if n == 0:
    return 0
  small, large = [], []
  r = int(math.isqrt(n))
  for i in range(1, r + 1):
    if n % i == 0:
      small.append(i)
      j = n // i
      if j != i:
        large.append(j)
  res = small + large[::-1]
  res.sort(reverse=True)
  return res


def unique(nums):
    i = 0
    c=len(nums)
    while i < c:
        j = i + 1
        while j < c:
            if nums[j] == nums[i]:
                nums.pop(j)
                c=len(nums)
            else:
                j += 1
        i += 1
    return nums


x=[1,2,3,4,1,2,3,4,1,1,1,3,4,1,2,2,2,4]
d=divisors(8).remove(8)
print(d)
```
